```c++
// sizeof(bool) is 1 by the C++ standard ([basic.types]): the value of the
// expression is exactly 1 byte.
cout << format("sizeof(bool) = {} byte\n", sizeof(bool));

// Still, 1 bit per bool IS available — via bit-packed storage.

// (a) std::bitset<N>: N bits fixed at compile time.
bitset<8> flags;
flags[0] = true; flags[7] = true;
cout << format("bitset<8> (bits 0 and 7 set) = {}\n", flags.to_string());

// (b) std::vector<bool>: dynamic, one bit per element.
const unsigned m = 1000000;
vector<bool> dense(m);
dense.assign(m, true); // all true
cout << format("vector<bool> of {} elements = {} bytes payload (packed to 1/8)\n",
              dense.size(), m / 8);

// (c) manual packing: four booleans sharing one byte.
unsigned char packed = 0;
packed |= 1u << 0; packed |= 1u << 2; packed |= 1u << 4; packed |= 1u << 6;
cout << format("4 booleans in one byte = {:08b}\n", (unsigned)packed);

// The price: a packed bool has no address of its own, and touching it is
// a read-modify-write of the whole word that holds it — which is exactly
// why a plain bool costs a whole byte.
```
