## Motivation

Recall the implementation of the class `Counter`:

````python
%%cpp
class Counter {
    int count=0;
public:
    Counter() = default;
    explicit Counter(int count) : count{count} {}
    int operator()() {
        return ++count;
    }
    void set_count(int count) { 
        if (count>=0) this->count=count; 
        else cerr << "Count cannot go negative.\n";  
    }
    operator int() const {
        return count;
    }
};

Counter counter1;
for (auto i=0; i<5; i++) cout << counter1() << ' '; cout << '\n';
counter1.set_count(10);
for (auto i=0; i<5; i++) cout << counter1() << ' '; cout << '\n';
````

We can modify the code as follows to count backwards:

````python
%%cpp
class Counter {
    int count;
public:
    Counter() = delete;
    explicit Counter(int count) : count{count} {}
    int operator()() {
        return set_count(count-1), count;
    }
    void set_count(int count) {
        if (count>=0) this->count=count; 
        else cerr << "Count cannot go negative.\n";  
    }
    operator int() const {
        return count;
    }
};

Counter counter2(3);
for (auto i=0; i<5; i++) cout << counter2() << ' '; cout << '\n';

// error: call to deleted constructor of 'Counter'
// Counter counter3; 
// Counter counter3();
````

Unlike the forward counter:

1. The backward counter has the default constructor deleted using the [`delete` keyword](https://en.cppreference.com/w/cpp/language/default_constructor.html#Deleted_default_constructor); and
2. the `operator()` calls `set_count` to decrement the `count`.

There is a slight inefficiency that `operator()()` needs to check if the new count is non-negative even when the step size is positive. If we want to implement more features specific for backward counter, we would have to worry about how to merge them with the forward counter. The complexity increases quickly when we want to introduce other kinds of counters later, and the code may not be backward compatible or efficient for the previous versions.

A better solution is to simply define different kinds of counters as separate derived classes of a common base class:
- Members of the base class can be [inherited](https://en.wikipedia.org/wiki/Inheritance_(object-oriented_programming)), allowing code reuse.
- Derived classes can override/overload members of its base classes, resulting in [polymorphism](https://en.wikipedia.org/wiki/Polymorphism_(computer_science)).

## Derived Class

### Inheritance

The following defines the forward counter as a base class:

````python
%%cpp
class Counter {
protected:
    int count = 0;
public:
    Counter() = default;
    explicit Counter(int count) : count{count} {}
    int operator()() {
        return ++count;
    };
    void set_count(int count) {
        if (count>=0) this->count=count; 
        else cerr << "Count cannot go negative.\n";
    }
    operator int() const {
        return count;
    }
};

Counter counter1;
for (auto i=0; i<5; i++) cout << counter1() << ' '; cout << '\n';
````

This is the same as the previous implementation except that `count` now has [`protected` member access](https://en.cppreference.com/w/cpp/language/access.html#Protected_member_access), which make it accessible to members and friends of derived classes.

```cpp
class Counter {
protected: // instead of private
    int count = 0;
public:
    ...
};
```

The following defines `BackwardCounter` as a subclass of the base class `Counter`:

````python
%%cpp
class BackwardCounter : public Counter {
public:
    BackwardCounter() = delete;
    explicit BackwardCounter(int count) : Counter(count) {}
    int operator()() {
        set_count(count - 1);
        return count;
    }
};

// BackwardCounter counter2; // error: call to deleted constructor of 'BackwardCounter'
BackwardCounter counter2(3);
for (auto i=0; i<5; i++) cout << counter2() << ' '; cout << '\n';
````

The class declaration 
```cpp
class BackwardCounter : public Counter { // declares Counter as a public base class
    ...
};
```
has a [base clause](https://en.cppreference.com/w/cpp/language/derived_class.html) that specifies the list of base classes.
- `public` member specifier ([public inheritance](https://en.cppreference.com/w/cpp/language/derived_class.html#Public_inheritance)) for a base class means the access to the members of the base class is also inherited.
- `protected` member specifier ([protected inheritance](https://en.cppreference.com/w/cpp/language/derived_class.html#Protected_inheritance)) means that `public` members of the base class become `protected`.
- `private`  member specifier ([private inheritance](https://en.cppreference.com/w/cpp/language/derived_class.html#Private_inheritance)) means that both the `public` and `protected` members of the base class become `private`.

The constructor `explicit BackwardCounter(int count) : Counter(count) {}` has a member initialization list that calls the base class's constructor `Counter(count)`. Note that [base classes are initialized before members](https://en.cppreference.com/w/cpp/language/initializer_list.html#Initialization_order).

To properly override a method even if there is no compile-time information about the actual type of the class, the method should be marked using the [`virtual` function specifier](https://en.cppreference.com/w/cpp/language/virtual.html) in the base class:

````python
%%cpp
class Counter {
protected:
    int count = 0;
public:
    Counter() = default;
    explicit Counter(int count) : count{count} {}
    virtual ~Counter() = default; // allow polymorphic deletion

    virtual int operator()() {  // can be overridden
        return ++count;
    }
    void set_count(int count) {
        if (count>=0) this->count=count; 
        else cerr << "Count cannot go negative.\n";
    }
    operator int() const {
        return count;
    }
};
````

::::{caution} Virtual destructor

For any class that is meant to be inherited, it is a good practice to make the destructor 
- public and virtual so that it can be properly overriden by the derived class to clean up the resources of the derived class; or
- protected and non-virtual so that the base class cannot be used for destruction.

::::{admonition} Hermes
:class: dropdown

A virtual function only works because the object already exists: dispatch happens at runtime by following the vtable pointer stored inside the object to find the right override — but a constructor runs on memory that has *just* been allocated, before any subobject or vtable pointer is in place (the constructor itself is what sets the vtable pointer to point at the class being built). So a "virtual constructor" would have to be dispatched through the very mechanism it is responsible for creating — a chicken-and-egg problem. There is also a sizing problem: `new Counter` must reserve exactly `sizeof(Counter)` bytes at compile time, so a runtime choice of "which" constructor runs could never change how much memory gets allocated; and unlike member functions, constructors are not inherited at all — `BackwardCounter` above defines its own `explicit BackwardCounter(int count)` that delegates to `Counter(count)` — so there is no base version for a derived constructor to override in the first place. For this reason C++ forbids it outright (g++ rejects the lecture's attempted line with `error: constructors cannot be declared 'virtual'`), and offers the two working alternatives instead: a virtual **destructor** — which *can* work because during destruction the object still fully exists with a valid vtable, hence `virtual ~Counter() = default;` above for safe polymorphic deletion — and **factory functions** (e.g. a static factory or abstract factory returning `unique_ptr<Counter>`) to achieve polymorphic *creation*.

It is optional but good practice to add the `override` qualifier to let the compiler know it is intended to override the base version. If the compiler cannot find a base version with the same signature, which may happen due to typos, the compiler will raise an error instead of creating a new overload.

Even though `*counter2` and `counter3` have type `Counter` instead of `BackwardCounter`, they both use `int operator()()` defined in `BackwardCounter`. This is achieved using [dynamic dispatch](https://en.wikipedia.org/wiki/Dynamic_dispatch) for virtual functions.

````python
%%hermes
Explain in a paragraph or two how C++ implements dynamic dispatch for 
virtual functions. Why the name virtual?
````

::::{admonition} Hermes
:class: dropdown

**How:** A virtual call is not compiled into a direct jump to a fixed address the way an ordinary (statically bound) call is. Instead, every class with at least one virtual function gets its own *vtable* — a per-class array of function pointers, one slot per virtual function — and every polymorphic object carries a hidden *vptr*, a pointer to its own class's vtable, which the constructor sets up. The emitted code for `(*counter2)()` above is therefore: load the object's vptr, index `operator()`'s slot in that table, and jump through the pointer found there. Since `counter1` is actually a `BackwardCounter`, its vptr points at `BackwardCounter`'s vtable, so both loops in the cell above end up calling `BackwardCounter::operator()` even though `counter2`/`counter3` have static type `Counter`. The price is one extra pointer load plus an indirect call per virtual call, and one hidden word per polymorphic object — I verified the layout with g++: two otherwise identical single-`int` classes measure `sizeof == 4` vs `sizeof == 16` once the method becomes `virtual` (8 bytes for the vptr, 4 more from alignment padding).

**Why "virtual":** because *which* function is invoked is deliberately left undecided at compile time and resolved at run time from the object's dynamic type. An ordinary function's target address is fixed when the program is built (static binding); a virtual function's target is only known once you look inside the live object, so its binding is "virtual" — dynamic/indirect rather than concrete. The name captures exactly this contrast with static dispatch: the compiler emits one generic calling routine and delegates the choice of actual function body to the run-time lookup through the object's vptr.

## Memory Management

### UndoRedoCounter

Consider implementing a counter that can undo and redo changes to its count. We need to modify `Counter` slightly to allow `set_count` to be overriden as well:

````python
%%cpp
class Counter {
protected:
    int count = 0;
public:
    Counter() = default;
    explicit Counter(int count) : count{count} {}
    virtual ~Counter() = default;

    virtual int operator()() {
        return ++count;
    }
    virtual bool set_count(int count) {
        if (count>=0) return this->count=count, true; 
        else return cerr << "Count cannot go negative.\n", false;
    }
    operator int() const {
        return count;
    }
};
````

The following is an implementation that uses a [circular buffer](https://en.wikipedia.org/wiki/Circular_buffer) to keep a history of count values:

````python
%%cpp
class UndoRedoCounter : public Counter {
    int *buffer;
    size_t historysize, head, tail, current;

    size_t after(size_t pos) const { return (pos+1) % historysize; }
    size_t before(size_t pos) const { return (historysize+pos-1) % historysize; }

    int push(int value) {
        if (head == before(tail)) tail=after(tail);
        return buffer[head = current = after(current)] = value;
    }

public:
    UndoRedoCounter() : UndoRedoCounter(0) {}
    explicit UndoRedoCounter(int count, int historysize=5)
    : Counter(count), historysize(historysize), head(0), tail(0), current(-1) {
        if (historysize<1) throw runtime_error("History size must be at least 1.");
        buffer = new int[historysize];
        push(count);
    }
    ~UndoRedoCounter() override {
        delete[] buffer;
    }

    int operator()() override {
        return push(Counter::operator()());
    }
    bool set_count(int count) override {
        return Counter::set_count(count) && (push(count), true);
    }

    bool can_undo() const { return current != tail; }
    bool can_redo() const { return current != head; }
    int undo() {
        if (can_undo()) current = before(current); else cerr << "Cannot undo.\n";
        return count=buffer[current];
    }
    int redo() {
        if (can_redo()) current = after(current); else cerr << "Cannot redo.\n";
        return count=buffer[current];
    }
    void print_history() const {
        cout << "History Buffer [size=" << historysize << "]:\n";
        for (size_t i = 0; i < historysize; ++i) {
            cout << "[" << i << "] " << buffer[i];
            if (i == head) cout << " <- head";
            if (i == tail) cout << " <- tail";
            if (i == current) cout << " <- current";
            cout << "\n";
        }
    }
};

UndoRedoCounter c;
c.print_history();
````
