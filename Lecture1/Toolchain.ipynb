If you really are taking this course, be warned that you are getting yourself into object-oriented programming (OOP). E.g., you will learn to define a class like the following:

````python
%%cpp
#include<string>
#include<iostream>

/**
 * @brief A CityUHK course.
 */
class Course {
private:
    std::string code; ///< The code of the course.
    std::optional<std::string> name; ///< The optional name of the course.

public:
    /**
     * @brief Constructs a new Course.
     *
     * @param code The course code.
     * @param name The optional course name (default to nullopt).
     */
    Course(std::string code, std::optional<std::string> name = std::nullopt)
        : code(std::move(code)), name(std::move(name)) { }

    /**
     * @brief Gets the course code.
     *
     * @return The course code.
     */
    inline const std::string& get_code() const {
        return code;
    }

    /**
     * @brief Sets the course code if the new code is not empty.
     *
     * @param new_code The new code to set.
     */
    inline void set_code(const std::string& new_code) {
        if (!new_code.empty()) { code = new_code; }
    }

    /**
     * @brief Gets the course name.
     *
     * @return The the course name.
     */
    inline const std::optional<std::string>& get_name() const {
        return name;
    }

    /**
     * @brief Sets the course name or removes it if the new name is empty.
     *
     * @param new_name The new course name to set.
     */
    inline void set_name(const std::string& new_name) {
        if (!new_name.empty()) {
            name = new_name;
        } else {
            name = std::nullopt;
        }
    }

    /**
     * @brief Gets the course full name (code bollowed by name).
     *
     * @return The course full name.
     */
    std::string get_full_name() const {
        return name.has_value() ? code + " " + name.value() : code;
    }

    /**
     * @brief Prints a welcome message for the course.
     */
    void say_welcome() const {
        std::cout << format("Welcome to {}!\n", this->get_full_name());
    }
};
````

The cleanest way to express "code required, name optional" in C++ is a **defaulted argument** (`name = ""`), treating an empty string as "no name." That same empty-string convention makes `set_name` trivial: assign the string, and an empty one naturally removes the name.

```cpp
#include <iostream>
#include <string>

class Course {
public:
    // code is required; name is optional (empty string means "no name").
    Course(std::string code, std::string name = "")
        : code(std::move(code)), name(std::move(name)) {}

    // "Welcome to {code} {name}!" when a name exists, "Welcome to {code}!" otherwise.
    void say_welcome() const {
        if (name.empty())
            std::cout << "Welcome to " << code << "!\n";
        else
            std::cout << "Welcome to " << code << " " << name << "!\n";
    }

    // Assign the new name; passing an empty string removes it.
    void set_name(const std::string& new_name) {
        name = new_name;
    }

private:
    std::string code;
    std::string name;
};

int main() {
    Course c1("CS2310");              // code only
    Course c2("CS2310", "Toolchain"); // code + name

    c1.say_welcome();   // Welcome to CS2310!
    c2.say_welcome();   // Welcome to CS2310 Toolchain!

    c1.set_name("Toolchain");  // add a name
    c1.say_welcome();          // Welcome to CS2310 Toolchain!

    c2.set_name("");           // remove the name
    c2.say_welcome();          // Welcome to CS2310!

    return 0;
}
```

Design notes:
- **Optional name via default argument** — `Course("CS2310")` and `Course("CS2310", "Toolchain")` both work, matching "providing their course code and, optionally, name."
- **Empty string = no name** — one shared sentinel for both the constructor and `set_name`, so "remove the course name if an empty string is given" is just an assignment; no special-casing needed.
- `say_welcome` is `const` (it only reads state), and `set_name` takes its argument by `const&`.

One alternative worth knowing: if you'd rather distinguish "no name given" from "name set to empty" as distinct states, use `std::optional<std::string>` for `name` instead of the empty-string sentinel — but for this spec the empty-string approach is simpler and reads better.

Note that the compiled code is much smaller than the original source code. We can use the [optimization flag `-O`](https://clang.llvm.org/docs/CommandGuide/clang.html#code-generation-options) to instruct the compiler to apply various optimization techniques to improve the generated code for different purposes:

````python
%%bash
cd private
clang++ --std=c++23 -S -O0 hello.ii -o hello_O0.s   # no optimization
clang++ --std=c++23 -S -O1 hello.ii -o hello_O1.s   # between -O0 and -O2
clang++ --std=c++23 -S -O2 hello.ii -o hello_O2.s   # moderate level
clang++ --std=c++23 -S -O3 hello.ii -o hello_O3.s   # maximum optimization, good for release
clang++ --std=c++23 -S -O3 -ffast-math hello.ii -o hello_Of.s  # aggressive, might violate standards
clang++ --std=c++23 -S -Og hello.ii -o hello_Og.s   # debugging
clang++ --std=c++23 -S -Os hello.ii -o hello_Os.s   # optimize for size
clang++ --std=c++23 -S -Oz hello.ii -o hello_Oz.s   # optimize for more size reduction
````

````python
%%bash
ls -la private/hello*.s
````

The two commonly used optimizations are:

- `-O3`, which is good for software release as it attempts to rearrange the code as much as needed to make the final executable run faster, but such rearrangements make it difficult to debug the source code because the mapping from the machine code to the source is not preserved.
- `-Og`, which is good for debugging as it minimizes rearrangements of the source and retains function names and variable scopes, allowing a debugger to step through the execution and show relevant information from the corresponding source code.

#### Step 3: Asssembler

The assembly code is further converted to a machine code.

````python
%%bash
cd private
clang++ -c hello.s -o hello.o
clang++ -c hello_O0.s -o hello_O0.o
clang++ -c hello_O1.s -o hello_O1.o
clang++ -c hello_O2.s -o hello_O2.o
clang++ -c hello_O3.s -o hello_O3.o
clang++ -c hello_Of.s -o hello_Of.o
clang++ -c hello_Og.s -o hello_Og.o
clang++ -c hello_Os.s -o hello_Os.o
clang++ -c hello_Oz.s -o hello_Oz.o
````

::::{admonition} Hermes
:class: dropdown

:::{note} What the flag does?
`clang++` is a *driver*: it orchestrates several separate programs (preprocessor, compiler, assembler, **linker**). For the final link step it has to pick some linker program to run. On Linux its default choice is GNU `ld` (the bfd linker bundled with binutils).

`-fuse-ld=lld` overrides that choice: it tells the driver to invoke **`lld`, the linker from LLVM**, instead.
:::


:::{note} Why it's used in this lecture?
1. **It's the matching piece of the LLVM toolchain.** The notebook compiles with `clang++` and links against `--stdlib=libc++` — i.e. the whole LLVM stack. `lld` is LLVM's counterpart to GNU `ld`, exactly as `clang` is to `gcc`. The course deliberately assembles a coherent toolchain: **Clang (compiler) + lld (linker) + libc++ (standard library)**.

2. **It makes the linker visible (pedagogical point).** Step 4 ("the linking process") is a separate stage done by a separate program. Passing `-fuse-ld=lld` explicitly exposes a choice that's usually invisible: the compiler driver *delegates* linking to some other binary. The exercise solution in cell `ca9f23fe` reinforces this — flags like `--std=c++23` matter at the compile/expand steps, while linking is the driver's job of running the linker.

3. **Practical advantages of lld over GNU `ld`:**
   - **Much faster** — lld runs in parallel and is significantly quicker, which matters on larger projects.
   - **Clearer diagnostics** — its linker errors are more readable (symbol names, file locations).
   - **Required for LLVM-specific features** — notably **LTO (link-time optimization)** only works end-to-end with lld, and it has better support for some modern ELF features.

So: the program would link fine without the flag (it'd just use GNU `ld` by default); the option is there both to complete the consistent LLVM toolchain story and to illustrate that "linking" is performed by a distinct, selectable tool.

#### Actually Portable Executable

Both `clang++` and `g++` are installable on nearly any platforms, but the executable compiled in one platform cannot run in another:

````python
%%hermes
Why an executable compiled by GCC in one platform may not run in another platform?
````

::::{admonition} Hermes
:class: dropdown

Because a compiled executable is a binary that "speaks" one platform's language. When g++ compiles, it produces machine code and a binary format **targeted at the specific OS + CPU the compiler is running on**. Portability breaks at three layers:

**1. CPU instruction set (ISA)**
The machine code in the executable is written for one architecture (x86-64, AArch64, RISC-V…). A CPU of a different architecture can't even *decode* those instructions. Example: a binary built on your Mac (Apple Silicon, AArch64) won't run on an Intel laptop.

**2. Executable file format & OS kernel**
Linux expects **ELF**, macOS expects **Mach-O**, Windows expects **PE/COFF**. The OS kernel/loader only understands its own format — drop a Linux ELF binary on a macOS machine and it's just an unreadable file. Even on the *same* ISA (Linux x86-64 vs Windows x86-64), this alone stops it.

**3. System interface & ABI**
Even where the file format matches, the binary assumes a specific system contract:
- **System calls** — the code calls the OS through the platform's syscall interface (Linux `read`/`write` vs macOS/BSD variants vs Windows `ReadFile`).
- **C/C++ runtime library** — it's linked against the platform's `glibc`/`libc++`/`MSVCRT`; those libraries exist in different versions and interfaces per platform.
- **ABI (calling convention, name mangling, struct layout, exception handling)** — e.g. the System V AMD64 ABI (Linux/macOS) differs from the Microsoft x64 ABI (Windows), so even the *way functions are called* is platform-specific.

It's not that GCC is "non-portable" — it's that **portability lives at the source level, not the binary level**:

- The **same C++ source** can be compiled by GCC on nearly any platform, but each run produces a binary for *that* platform. Hence "compile here, run there" doesn't apply to native executables — you must **recompile on (or for) the target platform** (cross-compilation lets one machine build for another).
- That's what the "Actually Portable Executable" section contrasts: truly portable binaries are the ones run by a uniform interpreter/runtime — e.g. **Java bytecode** (the JVM) or **WebAssembly** — where one artifact runs on any platform that provides the runtime, at the cost of an interpretation/translation layer.

So the answer to "why": the executable embeds decisions made at compile time (instruction set, file format, ABI, linked libraries) that are specific to the platform GCC was built and run for — a different platform's CPU or OS simply doesn't honor the same contract.

````python
%%hermes
Briefly list the exit codes for C++ programs.
````

:::::{admonition} Hermes
:class: dropdown

Briefly, the exit codes a C++ program can produce:

The basics
- **`0`** — success. Returned by `return 0;`, or **implicitly** if `main` reaches its end without a `return` statement.
- **Non-zero (any other value)** — indicates failure. The specific number is program-defined; by convention **`1`** is the generic "something went wrong".

Worth knowing
- **Only the low 8 bits are visible** — the shell (and `echo ${?}`) only ever sees the value modulo 256, so effectively exit codes are `0`–`255`.
- **`<cstdlib>` constants**: `std::exit(0)` / `std::exit(1)` mirror success/failure (POSIX systems also define `EX_OK` = 0 and `EX_FAIL` = 1).
- **`sysexits.h`** (POSIX) suggests specific non-zero codes for specific failure kinds: `EX_USAGE` (64, bad command-line usage), `EX_DATAERR` (65), `EX_NOINPUT` (66), `EX_IOERR` (74), `EX_SOFTWARE` (70), etc.
- **Signal terminations show up as `128 + signal number`** (when reported by a shell): e.g. **139** = 128 + 11 (SIGSEGV, segfault), **134** = 128 + 6 (SIGABRT, from `std::abort()`/failed assertions). Strictly, a program killed by a signal has no C++ exit code of its own — the runtime reports it.

Shell conventions (not set by the C++ program itself)
- **`126`** — command found but not executable
- **`127`** — command not found

So in practice for the notebook's purposes: `main` returns `0` for success, anything else (customarily `1`) for failure, and the shell exposes it via `${?}` — exactly what the cell earlier in the notebook demonstrates by running `./private/hello` and then `echo ${?}`.

- [C++ reference guide on `std::exit`.](https://en.cppreference.com/w/cpp/utility/program/exit)

````python
%%hermes
Is running `./hello` the same as running `hello` in bash?
````

:::::{admonition} Hermes
:class: dropdown
**No.** The difference is *how the shell resolves the command name*, and it changes which program actually runs — or whether one runs at all.

::::{note} The lookup rules

- **`./hello`** — the name contains a slash, so the shell treats it **as a path**: run the file `hello` in the **current working directory**. No `PATH` search happens at all.
- **`hello`** — the name contains no slash, so the shell does its normal lookup, in order:
  1. shell **builtin**
  2. shell **alias / function**
  3. search the directories in **`$PATH`** for an executable named `hello`

Crucially, **the current directory is not in `$PATH`** on modern systems (it used to be, years ago). So a bare `hello` will *not* find the `hello` you just compiled in your working directory — it fails with `hello: command not found` (exit code 127), whereas `./hello` runs fine.
::::
::::{note} Practical consequences

| | `./hello` | `hello` |
|---|---|---|
| Where it must live | current directory | any directory in `$PATH` (not `.`) |
| Not found | `No such file or directory` (127) | `command not found` (127) |
| Can run a shell function/builtin named `hello`? | no — it runs the file | **yes** — the function shadows your binary |
| Could run a *different* `hello` | no | yes — e.g. `/usr/bin/hello` if one exists |
| `argv[0]` the program sees | `./hello` | `hello` |

So the two are equivalent only in the narrow case where the file happens to live in a directory that *is* on `$PATH` (or you've enabled the non-default `set -P` in bash, which adds `.` to the search).
::::
::::{note} Why the notebook always writes `./...`

This is exactly why the lecture cells are written `cd private && ... ./hello_gcc` (cell 125) and `./private/hello` (cell 54): the `./` makes it explicit that **the binary you just built** is what runs, rather than relying on a `PATH` that (a) doesn't include the current directory anyway, and (b) might contain some unrelated program with the same name.

A small security footnote: keeping `.` out of `$PATH` is deliberate — otherwise, anyone who drops a malicious executable into a directory you `cd` into (a shared directory, a `/tmp` checkout, etc.) could have it run whenever you type a matching command. The `./` prefix sidesteps the problem entirely: you say precisely which file to run.

#### header files

A header file in C++ is a file with a `.h` or `.hpp` extension that contains declarations and definitions that can be shared across multiple source files.

````python
%%hermes
Why does C++ use header files? Are header files common in other programming languages?
````

:::::{admonition} Hermes
:class: dropdown

::::{note} Why C++ uses header files

Header files exist because of C++'s (inherited from C's) **per-translation-unit compilation model**: the compiler doesn't see your whole program at once. Each `.cpp` file is compiled *separately*, and the compiler only needs to know the **interfaces** (signatures) of things defined elsewhere — not their full code — in order to check calls and lay out objects.

So the workflow is:
1. `course.hpp` — **declarations** (what exists: function signatures, class shapes, constants) — shared with everyone.
2. `course.cpp` — **definitions** (the actual implementation) — compiled into `course.o`.
3. Any file that uses `Course` does `#include "course.hpp"` and the **preprocessor textually pastes the header in** (this is exactly what Step 1 of the notebook shows with `-E`).

Concrete reasons this is needed:

- **Type checking without the code** — the compiler can compile `main.cpp` that calls `course.say_welcome()` as long as it has seen the declaration; the linker later resolves the real body from `course.o`. This is what makes large, multi-file projects buildable (and lets you change an implementation without recompiling every file that uses it).
- **Single definition across translation units** — the *One Definition Rule* requires that, say, a class, be defined identically everywhere it's used. Pasting the same header into each `.cpp` is the mechanism that guarantees this consistency.
- **Templates and `inline`/`constexpr`** — these must have their full definition visible at every point of use, so they almost always live in headers (this is why *header-only libraries* exist — see the "Header-only Libraries" note earlier in the notebook — e.g. many Boost libraries).
- **Historically: it's just `#include`** — the preprocessor treats it as text substitution. That's why you've seen the downsides: include guards / `#pragma once`, and slow compilation from giant header chains.
::::

::::{note} Are header files common in other languages?

**Rare — C++ is an outlier among major modern languages.** The split it does (separate declaration files + preprocessor pasting + per-file compilation + linker) is mostly unique to the C/C++ family:

| Language | Mechanism instead of headers |
|---|---|
| **C** | Same header model as C++ (`.h` + `#include`) — it's where the idea comes from |
| **Rust** | `.rs` modules; `mod`/`pub` control visibility, everything compiled together by `rustc` — one `mod.rs` per module |
| **Java / C#** | One public type per file (convention/enforced), visibility via `public`/`private`, package namespaces; no preprocessing or manual splitting |
| **Go** | All `.go` files in a package are compiled together; visibility = capitalized or not |
| **Python / JS** | `import` statements; the file *is* the interface — no separate declaration file |

So the *concept* behind headers — separating "what a thing is" from "how it works" — is universal (it's literally called an **interface** in most other languages), but the *mechanism* of pasting declaration files via a preprocessor is characteristic of C/C++. If you ever see "header file" outside C/C++, it's usually just a filename convention (e.g. `.h` in C, or "header" meaning the first lines of a file, like HTTP/Unix socket headers), not the same compilation construct.

#### binary file

Hex editors such as [radare2](https://github.com/radareorg/radare2?tab=readme-ov-file#radare2-libre-reversing-framework-for-unix-geeks) can help us read and edit machine code.

````python
%%hermes
Give a brief but concrete example how one can reverse engineer a software using a hex editor.
````

:::::{admonition} Hermes
:class: dropdown

::::{note} Example: find and patch the message inside `private/hello`

**1. Inspect — spot the source string hiding in the binary.**

The program was compiled from `auto who = "World"; std::cout << std::format("Hello, {}!\n", who);`. Open the executable in a hex editor — e.g.

```bash
r2 -q -c 'izz' private/hello     # list all embedded strings + their offsets
```

You'll see `World` (and `Hello, {}!\n`) listed, sitting in the read-only data region (`.rodata`). Just by *reading bytes*, you've already reverse-engineered the program's behavior: it prints a formatted greeting, and the constant `"World"` lives at some offset — say `0x222e`.

**2. Look — see the raw bytes.**

Jump to that offset. A hex editor shows you exactly what the compiler stored:

```
Offset:      2220
Hex:   48 65 6c 6c 6f 2c 20 7b 7d 21 0a 00 57 6f 72 6c 64 00
ASCII: H  e  l  l  o  ,    {  }  !  \n \0 W  o  r  l  d  \0
```

Two NUL-terminated C strings side by side. A hex editor is just a way to see this layer — the executable is mostly machine code (binary, opaque) interleaved with these readable constants.

**3. Edit — patch a byte, no recompilation.**

Overwrite the 5 bytes `57 6f 72 6c 64` (`World`) with `59 6f 75 00 00` (`You\0\0`). (In radare2: `s 0x222e` then `wx 596f750000`.) Save.

**4. Verify — the program's behavior changed.**

```bash
$ ./private/hello
Hello, You!
```

Same binary, different behavior — you changed what the CPU does by editing 5 bytes. That's the essence of reverse engineering at the hex level: **observe → locate → modify → verify**, with no source code involved.
::::

::::{note} Why this works, and its limits

- **Strings survive compilation untouched** — the compiler just copies `"World\0"` into the binary's data section, so ASCII search in any hex editor finds it instantly. This is how people find product names, URLs, error messages, or license keys in software.
- **Rules of thumb for patching:** replace with a string of the *same or shorter* length (the null terminator ends it; extra old bytes after the new `\0` are ignored — that's why `You\0\0` worked). Longer replacements would clobber adjacent data.
- **You can also patch *code*, not just data** — e.g. flip a single byte of a `cmp`/`jz` pair to make a license check always pass (classic "cracking" exercise) — but that requires disassembling first, which is where tools like radare2 pull double duty as disassemblers: raw hex for data, decoded assembly for logic.
- **Practical caveats:** the file must stay writable/executable (`-rwxr-xr-x` — see the permission-string section above) after saving; and real shipping software fights this with code signing, encryption, and packing, so the hex-level tricks above are a teaching demo, not a general cracking methodology.

#### static library

A static library in C++ is a collection of object files combined into a single archive file (typically with a .a extension), which is included directly into the executable at compile time.

````python
%%hermes
When is it preferrable to use a static library instead of a dynamic library, or 
the object files directly.
````

:::::{admonition} Hermes
:class: dropdown

Quick recap of the three forms, then the decision rules:

- **Object files** (`hello.o`) — what the assembler emits; linked directly.
- **Static library** (`libfoo.a`) — an *archive* of object files with an index; the linker pulls out only the members it needs.
- **Dynamic library** (`libfoo.so`) — a shared object loaded at runtime by the dynamic linker (the `libc++.so`/`libstdc++.so` you listed with `ldconfig -p` above).

::::{note} When to use **object files directly**

- A single, self-contained program with a small number of translation units — exactly what Step 4 of the notebook does (`clang++ ... hello.o -o hello_again`).
- The "library" *is* the program (you're not reusing the code elsewhere), so there's nothing to gain from archiving.
- You need the most control over exactly what gets linked, or you're about to use LTO (object files + LTO give the optimizer the whole picture).

Caveat: the linker includes an object file **wholesale** once any of its symbols is needed — unused functions inside it stay in the binary (unless you add `--gc-sections`/`-ffunction-sections` for link-time garbage collection).
::::
::::{note} When a **static library** beats object files

- **Distribution / reuse**: one artifact (`libfoo.a`) instead of shipping a pile of `.o` files with naming/build-order headaches.
- **Selective linking**: the archive's index lets the linker pull in *only the members that resolve missing symbols* — so a consumer using 5 functions of a 50-file library embeds 5 members, not all 50. That's the main mechanical advantage over raw object files.
- **Many consumers, one codebase**: every program links its own copy, which leads directly to the next point.
::::
::::{note} When a **static library** beats a **dynamic library**

| Concern | Static wins because… |
|---|---|
| **Self-contained deployment** | No runtime dependency on the `.so` being present, findable, and the right version — no `LD_LIBRARY_PATH`/`rpath` fuss, no "works on my machine" ABI breakage at startup. |
| **Embedded / minimal targets** | The environment may have no dynamic loader at all (microcontrollers, static-RTOS builds). |
| **Reproducible builds** | The exact code is frozen into the binary; behavior doesn't depend on what's installed at runtime. |
| **No public ABI to maintain** | You don't need to preserve a stable exported interface across versions; a fix ships inside the rebuilt executable. |
| **Symbol control** | The implementation isn't an importable interface (though the symbols still sit in the binary). |

Costs of static: each executable embeds its own copy (bigger binaries; the notebook already showed how much bigger executables get), a bug fix requires rebuilding and redistributing *every* consumer, and if two statically-linked components embed the *same* library, a process can end up with **two copies** of its global state — the classic "two libstdc++ in one process" problem, which is exactly why the C++ standard library is normally linked dynamically (the `libc++.so` in cell `cf648cf0`).
::::
::::{note} Rule of thumb

- **One program, few files** → link the object files directly (notebook's `hello_again`).
- **Code reused by several programs, or distributed as a deliverable, and you want a single self-contained executable** → static archive.
- **Code shared across many programs/processes where memory, hot-patching, or plugin loading matter** → dynamic library.
