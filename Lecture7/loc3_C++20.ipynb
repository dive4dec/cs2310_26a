```cpp
#include <iostream>
#include <format>
#include <vector>
```

```cpp
using std::cout, std::format, std::vector;

auto &loc3(auto &&x) {
    cout << format("@{:p}\n", static_cast<const void *>(&x));
    return x;
}
```

```cpp
{
    auto u(loc3(vector<int>{1, 2, 3}));
    loc3(u).push_back(4);
}
```

```cpp
loc3(1) = 0;   // works?!
```
