```c++
/** Count the symbols of a null-terminated UTF-8 C string.
 * A symbol starts at every byte that is not a continuation byte (0b10xxxxxx). */
size_t utf8_symbols(const char* s) {
    size_t n = 0;
    while (*s) {
        if ((static_cast<unsigned char>(*s) & 0xC0) != 0x80) n++;
        s++;
    }
    return n;
}
```

```c++
/** Count the symbols of a null-terminated UTF-16 C string.
 * A high surrogate (0xD800-0xDBFF) is followed by the low surrogate of the same symbol. */
size_t utf16_symbols(const char16_t* s) {
    size_t n = 0;
    while (*s) {
        n++;
        if (0xD800 <= *s && *s <= 0xDBFF) s++;
        s++;
    }
    return n;
}
```

```c++
cout << format("\"15\": bytes={}, symbols={}\n", strlen("15"), utf8_symbols("15"));
cout << format("\"思\": bytes={}, symbols={}\n", strlen("思"), utf8_symbols("思"));
cout << format("\"😎\": bytes={}, symbols={}\n", strlen("😎"), utf8_symbols("😎"));
cout << format("u\"思😎\": code units={}, symbols={}\n", sizeof(u"思😎") / 2 - 1, utf16_symbols(u"思😎"));
```
